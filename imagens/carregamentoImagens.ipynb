```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
```

```python
urls = ["https://blogs.nvidia.com/wp-content/uploads/2022/02/mountain-lake-reflection-2-scaled.jpg",
        "https://cdn.britannica.com/84/158184-050-1D7ADEB5/balloon.jpg",
        "https://cdn.hswstatic.com/gif/airplane-windows.jpg"]

for url in urls:
    image = io.imread(url)
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
    final_frame = cv.hconcat((image, image_2))
    cv2_imshow(final_frame)
    print(image_2.shape, '\n\n')


    
```

```python
# Desebgar yn retâgulo preenchido da cor preta no canto superior esquerdo da imagem
image_2[0:100, 0:200, 1] = 0

cv2_imshow(image_2)
```

```python
# Aumentando o brilho da imagem por um fator de 1.5
image_3 = image_2 * 1.5
cv2_imshow(image_3)

# Dividindo o brilho da imagem por um fator de 1.5
image_3 = image_2/1.5
cv2_imshow(image_3)
```

```python
# Calcula a média dos canais de cor (RGB) para cada pixel
gray_image_mean = image_2.mean(axis=2).astype(image_2.dtype)

cv2_imshow(gray_image_mean)
print(f'Dimensões da imagem em tons de cinza (média dos canais): {gray_image_mean.shape}')
```
